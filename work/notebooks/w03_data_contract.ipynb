## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

This section outlines the inherent limitations of the data, highlighting what it cannot reveal or accurately represent, and potential issues that might arise during analysis or modeling.

### Potential Data Limitations:

1.  **Unbalanced History/Data Skew:**
    *   The current data set (a single month) provides a limited snapshot. We cannot infer long-term trends or seasonal patterns reliably without more historical data.
    *   The number of clients or content items with available GSC/GA4 data might be unbalanced, meaning some clients or content pieces have more comprehensive tracking than others. This can lead to biased insights if not accounted for.
    *   Some `ai_` related metrics and `scroll_events` show many missing values, suggesting these features might not be consistently collected or available for all content/clients. This imbalance limits their use for comprehensive analysis or as reliable predictors.

2.  **GSC-only Early Rows / GA4 Adoption:**
    *   The `client_has_gsc` and `client_has_ga4` columns, along with `gsc_data_available` and `ga4_data_available`, indicate that not all clients have both data sources. There might be a historical bias where older clients or content only have GSC data, or GA4 adoption is not universal. This means we cannot always compare performance across all metrics for all entities.
    *   The `ga4_data_available` column is of `object` type and needs further inspection; it might contain mixed types or strings that require cleaning before it can be reliably used as a boolean indicator.

3.  **Window Overlaps/Continuity:**
    *   While the `report_date` within the loaded file is continuous for March 2026, we've only loaded one month. If we were to load multiple months, we would need to explicitly verify that there are no gaps or overlaps between the monthly files (e.g., if one `month=X` file ends on the 25th and `month=Y` starts on the 5th of the next month).
    *   The data only contains daily aggregates. Finer-grained temporal insights (e.g., hourly performance) are not available.

4.  **Absence of External Factors:**
    *   The data solely focuses on content performance metrics and client/content identifiers. It does not include external factors that could significantly impact performance, such as marketing campaigns, industry news, competitor activities, holidays, or algorithm updates. Without this context, any observed changes in performance cannot be fully attributed to internal content characteristics alone.

5.  **Limited Client/Content Context:**
    *   We have `client_hash_id` and `content_hash_id` but no rich metadata about the clients (e.g., industry, size) or the content (e.g., topic, format, length, target audience). This limits our ability to understand *why* certain content or clients perform differently, beyond the raw metrics.

### 4.2 Imputing Missing Values for Numerical Metrics

For numerical columns where missing values likely indicate no activity (e.g., `ga4_pageviews`, `sessions_organic`, `ai_chatgpt`, `scroll_events`), we will impute `NaN` with `0`. This is a common practice for count-based metrics when missingness implies an absence of the measured event.

For `gsc_avg_position`, where `NaN` could mean no impressions or no ranking, imputing with `0` is a reasonable initial approach, signifying an undefined or non-existent average position.

```python
columns_to_impute_with_zero = [
    'ga4_pageviews',
    'ga4_sessions',
    'ga4_users',
    'ga4_engaged_sessions',
    'ga4_total_engagement_sec',
    'sessions_organic',
    'sessions_direct',
    'sessions_referral',
    'sessions_social',
    'sessions_paid',
    'sessions_ai',
    'ai_chatgpt',
    'ai_perplexity',
    'ai_gemini',
    'ai_copilot',
    'ai_claude',
    'ai_meta',
    'ai_other',
    'scroll_events'
]

for col in columns_to_impute_with_zero:
    if col in df.columns:
        df[col] = df[col].fillna(0)

# Impute gsc_avg_position with 0 as well, assuming missing implies no ranking/impressions
if 'gsc_avg_position' in df.columns:
    df['gsc_avg_position'] = df['gsc_avg_position'].fillna(0)

print("Missing values after imputation:")
display(df.isnull().sum()[df.isnull().sum() > 0])
```
